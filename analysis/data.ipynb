```python
import pandas as pd
```

```python
class DOMAINS:
    overall: str = "dz_simd_rank"
    income: str = "dz_simd_income_rank"
    employment: str = "dz_simd_employment_rank"
    education: str = "dz_simd_education_rank"
    health: str = "dz_simd_health_rank"
    access: str = "dz_simd_access_rank"
    crime: str = "dz_simd_crime_rank"
    housing: str = "dz_simd_housing_rank"

df = pd.read_parquet(
    "../data/processed/scotland_clustering_analysis_dataset.parquet",
    columns=[
        "sequence_id", "collection_date", "who_voc",
        "datazone", "dz_simd_rank", "dz_simd_income_rank",
        "dz_simd_employment_rank", "dz_simd_education_rank", "dz_simd_health_rank",
        "dz_simd_access_rank", "dz_simd_crime_rank", "dz_simd_housing_rank"
    ]
)
```
